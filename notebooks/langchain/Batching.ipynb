```python
%load_ext dotenv
%dotenv
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
```

```python
chat_template = ChatPromptTemplate.from_messages([
    ('human', "I've recently adopted a {pet} which a {breed}. Could you suggest serveral training tips?"),
])
```

```python
chat_value = chat_template.invoke({'pet': 'dog', 'breed': 'shepherd'})
```

```python
chat_value
```

```python
chat = ChatOpenAI(
    model='gpt-4',
    temperature=0,
    max_tokens=250,
    seed=365
)
```

```python
chain = chat_template | chat
```

```python
%%time
chain.batch([
    {'pet': 'dog', 'breed': 'shepherd'},
    {'pet': 'dragon', 'breed': 'night furry'}
])
```

```python
%%time
chain.invoke({'pet': 'dog', 'breed': 'shepherd'})
```

```python
%%time
chain.invoke({'pet': 'dragon', 'breed': 'night furry'})
```
