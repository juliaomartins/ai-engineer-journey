```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableParallel
```

```python
chat_template_books = ChatPromptTemplate.from_template('''
Suggest three of the best intermediate-level {programming language} books.
ANswer only by listing the books.
''')
```

```python
chat_template_projects = ChatPromptTemplate.from_template('''
Suggest three interesting {programming language} projects suitable for intermediate-level programmers.
Answer only by listing the projects.
''')
```

```python
chat_template_time = ChatPromptTemplate.from_template('''
I'm an itermediate level programmer.

consdier the following literature:
{books}

Also, consider the following projects:
{projects}

Roughly how much time would it take me to complete the literature and the projects?
''')
```

```python
chat = ChatOpenAI(
    model='gpt-4',
    temperature=0,
    max_tokens=500,
    seed=365
)
```

```python
string_parser = StrOutputParser()
```

```python
chain_books = chat_template_books | chat | string_parser
```

```python
chain_projects = chat_template_projects | chat | string_parser
```

```python
chain_parallel = RunnableParallel({'books': chain_books, 'projects': chain_books})
```

```python
response = chain_parallel.invoke({'programming language': 'python'})
```

```python
print(response)
```

```python
chain_parallel.get_graph().print_ascii()
```

```python
chain_time_1 = (RunnableParallel({'books': chain_books, 'projects': chain_books})
              | chat_template_time
              | chat
              | string_parser)
```

```python
chain_time_2 = ({'books': chain_books, 'projects': chain_books}
              | chat_template_time
              | chat
              | string_parser)
```

```python
print(chain_time_2.invoke({'programming language': 'python'}))
```
