```python
%load_ext dotenv
%dotenv
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough
```

```python
chat_prompt_tools = ChatPromptTemplate.from_template('''
What are the five most important tools a {job title} needs?
Answer only by listing the tools.''')
```

```python
chat_prompt_strategy = ChatPromptTemplate.from_template('''
Consdireing the tools Provided, develop a strategy for effectively learning and mastering them:
{tools}''')
```

```python
chat = ChatOpenAI(
    model='gpt-4',
    temperature=0,
    max_tokens=250,
    seed=365
)
```

```python
str_output_parser = StrOutputParser()
```

```python
chain_tools = chat_prompt_tools | chat | str_output_parser | {'tools': RunnablePassthrough()}
```

```python
response_tools = chain_tools.invoke({'job title': 'Data Scientist'})
```

```python
print(response_tools)
```

```python
chain_strategy = chat_prompt_strategy | chat | str_output_parser
```

```python
response_strategy = chain_strategy.invoke({'tools': '''
1. Python
2. R Programming
3. SQL
4. Tableau
5. Hadoop
'''})
```

```python
print(response_strategy)
```

```python
chain_long =  chat_prompt_tools | chat | str_output_parser | {'tools': RunnablePassthrough()} | chat_prompt_strategy | chat | str_output_parser
```

```python
response_long = chain_long.invoke({'job title': 'Software Developer'})
```

```python
print(response_long)
```
