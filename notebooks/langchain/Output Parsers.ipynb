```python
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage
from langchain_classic.output_parsers import DatetimeOutputParser
from langchain_core.output_parsers import (StrOutputParser, CommaSeparatedListOutputParser)
```

```python
message_h = HumanMessage(content=''' Can you tell fun fact probably I didn't know. ''')
```

```python
llm = ChatOpenAI(
    model='gpt-4',
    temperature=0,
    max_tokens=250,
    seed=365
)
```

```python
response = llm.invoke([message_h])
```

```python
response
```

# StrOutputParser()

```python
str_output_parser = StrOutputParser()
```

```python
response_parser = str_output_parser.invoke(response)
```

```python
response_parser
```

```python
print(response_parser)
```
