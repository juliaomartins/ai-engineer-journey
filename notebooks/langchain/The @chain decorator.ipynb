### The @chain decorator

```python
from langchain_core.runnables import RunnableLambda
from langchain_core.runnables import chain
```

```python
def find_sum(x):
    return sum(x)


def find_square(x):
    return x**2
```

```python
runnable_sum = RunnableLambda(find_sum)
runnable_square = RunnableLambda(find_square)
```

```python
runnable_sum.invoke([1, 2, 5])
```

```python
runnable_square.invoke(8)
```

```python
chain_1 = runnable_sum | runnable_square
```

```python
chain_1.invoke([1, 2, 5])
```

```python
@chain
def find_sum(x):
    return sum(x)


@chain
def find_square(x):
    return x**2
```

```python
chain_2 = find_sum | find_square
```

```python
chain_2.invoke([1, 2, 5])
```

```python
chain_2.get_graph().print_ascii()
```
