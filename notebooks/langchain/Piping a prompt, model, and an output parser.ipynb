```python
%load_ext dotenv
%dotenv
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import CommaSeparatedListOutputParser
```

```python
list_instructions = CommaSeparatedListOutputParser().get_format_instructions()
```

```python
chat_template = ChatPromptTemplate.from_messages([
    ('human', "I've recently adopted a {pet}. Could you suggest three {pet} names?\n" + list_instructions)
])
```

```python
chat_template
```

```python
print(chat_template.messages[0].prompt.template)
```

```python
chat = ChatOpenAI(
    model='gpt-4',
    temperature=0,
    max_tokens=250,
    seed=365
)
```

```python
chat_value = chat_template.invoke({'pet': 'dog'})
```

```python
response = chat.invoke(chat_value)
```

```python
list_output_parser = CommaSeparatedListOutputParser()
```

```python
response_parser = list_output_parser.invoke(response)
```

```python
response_parser
```

```python
chain = chat_template | chat | list_output_parser
```

```python
chain.invoke({'pet': 'cat'})
```

```python
chain.invoke({'pet': 'dragon'})
```
